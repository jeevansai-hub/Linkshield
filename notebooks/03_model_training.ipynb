# Notebook 03: Model Training & Hyperparameter Setup

**Project**: LinkSentinel (`LinkShield`)
**Objective**: Train Model 1 (Logistic Regression with StandardScaler pipeline) and Model 2 (Random Forest Classifier on unscaled features) across Experiment A (UCI Baseline) and Experiment B (LinkSentinel Raw URL Engine) with seed `42`.

```python
import os
import sys
import joblib
import pandas as pd
import numpy as np

sys.path.append('..')
from src.models.train_evaluate import ModelPipeline

# Load processed splits
splits = joblib.load('../data/processed/splits.joblib')
print("Loaded train/validation/test splits successfully.")
```

## 1. Train Experiment A Models (UCI Phishing Baseline)

```python
X_uci_tr, y_uci_tr = splits['uci']['X_train'], splits['uci']['y_train']
X_uci_val, y_uci_val = splits['uci']['X_val'], splits['uci']['y_val']

# Logistic Regression Pipeline
lr_uci = ModelPipeline(model_type='logistic_regression', random_state=42)
lr_uci.fit(X_uci_tr, y_uci_tr)
val_metrics_lr_uci = lr_uci.evaluate(X_uci_val, y_uci_val)

# Random Forest Classifier
rf_uci = ModelPipeline(model_type='random_forest', random_state=42)
rf_uci.fit(X_uci_tr, y_uci_tr)
val_metrics_rf_uci = rf_uci.evaluate(X_uci_val, y_uci_val)

print("UCI Baseline Validation Metrics:")
print("  Logistic Regression F1:", val_metrics_lr_uci['f1_score'])
print("  Random Forest F1:       ", val_metrics_rf_uci['f1_score'])
```

## 2. Train Experiment B Models (LinkSentinel Raw URL Engine)

```python
X_url_tr, y_url_tr = splits['url_engine']['X_train'], splits['url_engine']['y_train']
X_url_val, y_url_val = splits['url_engine']['X_val'], splits['url_engine']['y_val']

# Logistic Regression Pipeline
lr_engine = ModelPipeline(model_type='logistic_regression', random_state=42)
lr_engine.fit(X_url_tr, y_url_tr)
val_metrics_lr_engine = lr_engine.evaluate(X_url_val, y_url_val)

# Random Forest Classifier
rf_engine = ModelPipeline(model_type='random_forest', random_state=42)
rf_engine.fit(X_url_tr, y_url_tr)
val_metrics_rf_engine = rf_engine.evaluate(X_url_val, y_url_val)

print("\nLinkSentinel Raw URL Engine Validation Metrics:")
print("  Logistic Regression F1:", val_metrics_lr_engine['f1_score'])
print("  Random Forest F1:       ", val_metrics_rf_engine['f1_score'])
```

## 3. Save Model Artifacts

```python
os.makedirs('../models', exist_ok=True)
model_artifacts = {
    'uci_lr': lr_uci,
    'uci_rf': rf_uci,
    'engine_lr': lr_engine,
    'engine_rf': rf_engine,
    'feature_names': splits['url_engine']['feature_names']
}
joblib.dump(model_artifacts, '../models/linksentinel_models.joblib')
print("Trained model artifacts saved to models/linksentinel_models.joblib.")
```
